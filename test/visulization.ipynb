```python
import sys
sys.path.append('./src')
import KGBN
```

```python
# visulize the Palma model
network_file_no_phenotype = '../input_files/Palma2021_no_phenotype.txt' # no phenotype nodes
network = KGBN.load_network(network_file_no_phenotype)
# KGBN.vis_network(network, output_html="Palma2021_nophe.html", interactive=True, color_node='lightblue', physics=False)
```

```python
# extended PBN by adding drug targets
drug_targets = ['BCL2','SYK', 'BTK','MAP2K1','MAP2K2','KIT']
genes = list(network.nodeDict.keys())
genes = genes + drug_targets
KG_string, relations = KGBN.load_signor_network(genes, joiner='&',saveto='KG_evidence.txt')
KG = KGBN.load_network_from_string(KG_string)
# KGBN.vis_network(KG, output_html="drug_targets_KG.html", interactive=True, color_node='#E1F2D0', physics=False)

# Visualize extended PBN
from KGBN.vis import vis_extension
pbn_string = KGBN.merge_networks([network, KG], method="PBN", prob=0.6)
pbn = KGBN.load_pbn_from_string(pbn_string)
# vis_extension(network,pbn, output_html="Palma2021+KG.html", interactive=True, color_node='lightblue', color_edge='lightblue', extension_color_node='#FBE1BE', physics=False)
```

```python
# extended PBN by adding drug targets plus additional mutations
mutations = ['NRAS', 'DAB2IP']
genes = list(network.nodeDict.keys())
genes = genes + mutations + drug_targets
KG_string_mutations, relations_mutations = KGBN.load_signor_network(genes, joiner='&',saveto='KG_mutations_evidence.txt')
KG_mutations = KGBN.load_network_from_string(KG_string_mutations)
pbn_string_mutations = KGBN.merge_networks([network, KG_mutations], method="PBN", prob=0.6)
pbn_mutations = KGBN.load_pbn_from_string(pbn_string_mutations)
# vis_extension(network,pbn_mutations, output_html="Palma2021+KG_mutations.html", interactive=True, color_node='lightblue', color_edge='lightblue', extension_color_node='#FBE1BE', physics=False)
```

```python
edge_scores_mutations = {(src, tgt): score
               for src, tgt, _, score in relations_mutations
               if score is not None}
print(edge_scores_mutations)
KGBN.vis_extension(network, pbn_mutations,
                   interactive=False,
                   layout='shell',
                   figsize=(18, 7),
                   color_node='#AED6F1',
                   extension_color_node='#AED6F1',
                   node_groups=
                   {'New Mutations': ({'NRAS', 'DAB2IP'}, '#D783FF')},
                   edge_scores=edge_scores_mutations,
                   )
```
