# Compress network

This notebook demonstrates how to use the model compression functionality
to simplify Boolean networks before optimization by:
1. Finding and removing non-observable and non-controllable nodes
2. Collapsing linear paths that don't provide additional information
3. Visualizing compression results

```python
import sys
import os
import numpy as np
import KGBN
 
```

```python
network_string = """
Input1 = Input1
Input2 = Input2
A = Input1
B = A
C = B
D = C | Input2
E = D
Output1 = E
F = Input1 & Input2
G = F
Isolated = Isolated
H = G
Output2 = H
SelfLoop = SelfLoop & Input1
"""
measured_nodes = {'Output1', 'Output2'}
perturbed_nodes = {'Input1', 'Input2'}

network = KGBN.load_network_from_string(network_string)
```

```python
KGBN.vis_network(network)
```

```python
KGBN.vis_network(network, output_html="files/original_network.html", interactive=True)
```

```python
# Compress
compressed_network, compression_info = KGBN.compress_model(
    network,
    measured_nodes=measured_nodes,
    perturbed_nodes=perturbed_nodes
)
```

```python
KGBN.vis_compression(
    network,
    compressed_network,
    compression_info
    )
```

```python
KGBN.vis_compression(
    network,
    compressed_network,
    compression_info,
    "files/compression_comparison.html",
    interactive=True
)
```

```python
KGBN.vis_network(compressed_network)
```

```python
print("Compressed network:")
print("\n".join(compressed_network.equations))
```
